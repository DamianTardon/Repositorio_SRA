```python
# =============================================================================
# Librerias
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

```python
# =============================================================================
# Parametros del sistema
# =============================================================================
Ad0 = 1e5            # Ganancia DC del VFA (veces)
fp1_VFA = 10         # Primer polo del VFA (Hz)
fp2_VFA = 5.06e6     # Segundo polo del VFA (Hz)
fg = 2e6             # Frecuencia de ancho de banda potencial (Hz)

# =============================================================================
# Analisis del Compensador
# =============================================================================
Rc1 = 1e3                               # Resistencia de la red compensadora (Ohm)
Rc2 = 1e3                               # Resistencia de la red compensadora (Ohm)
Cc = 1 / (2 * np.pi * fp2_VFA * Rc1)    # Capacitancia de la red compensadora (F)

# =============================================================================
# Analisis del amplificador CFA
# =============================================================================
AVf_CFA_DC = 21.5                   # Ganancia en DC del CFA (veces)
R2 = 995                            # Resistencia de realimentacion del CFA (Ohm)
R1 = R2 / (2 * AVf_CFA_DC - 1)      # Resistencia de realimentacion del CFA (Ohm)

# =============================================================================
# Analisis del amplificador compuesto
# =============================================================================
fp_CFA = 33.3e6                     # Frecuencia de polo del CFA (Hz)
fp_C = 2 * fp2_VFA                  # Frecuencia de polo del compensador (Hz)
M_phi = 180 - np.degrees(np.arctan(fg / fp1_VFA) - np.arctan(fg / fp_CFA) - np.arctan(fg / fp_C)) # Margen de fase del sistema compuesto (grados)

# =============================================================================
# Visualizacion de resultados
# =============================================================================
print("="*50)
print("Analisis del Compensador")
print("-"*50)
print(f"Rc1= {Rc1/1e3:.0f} kΩ")
print(f"Rc2= {Rc2/1e3:.0f} kΩ")
print(f"Cc= {Cc*1e12:.0f} pF")
print(f"Polo fp_C= {fp_C/1e6:.2f} MHz")
print("="*50)
print("Analisis del amplificador CFA")
print("-"*50)
print(f"R1= {R1:.0f} Ω")
print(f"R2= {R2:.0f} Ω")
print("="*50)
print("Analisis del sistema compuesto")
print("-"*50)
print(f"Margen de fase M_phi= {M_phi:.2f}°")
```

```python
# =============================================================================
# Respuesta al escalon
# =============================================================================
V_pico = 115.72e-3
V_regimen = 97.82e-3

OS = (V_pico - V_regimen) / V_regimen * 100     # Sobrepaso (overshoot) porcentual

# Factor de amortiguamiento
ln_os = np.log(OS / 100)
zeta = -ln_os / np.sqrt(np.pi**2 + ln_os**2)

# Margen de fase (grados)
Mp = np.arctan((2 * zeta) / np.sqrt(np.sqrt(1 + 4 * zeta**4) - 2 * zeta**2)) * (180 / np.pi)

print("="*50)
print("Respuesta al escalon")
print("-"*50)
print(f"Sobrepaso (OS) = {OS:.2f} %")
print(f"Zeta = {zeta:.3f}")
print(f"Margen de fase (Mp) = {Mp:.2f}°")
```
