```python
# =============================================================================
# Librerias
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

```python
# =============================================================================
# Parametros del sistema
# =============================================================================
Ad0 = 1e5            # Ganancia DC del VFA (veces)
fp1_VFA = 10         # Primer polo del VFA (Hz)
fp2_VFA = 5.06e6     # Segundo polo del VFA (Hz)
fg = 2e6             # Frecuencia de ancho de banda potencial (Hz)
M_phi = 65           # Margen de fase (grados)
K = 0.1              # Factor de realimentacion
CT = 4.8e-12         # Capacitancia del CFA (F)

# =============================================================================
# Analisis del amplificador VFA
# =============================================================================
# Magnitud y Fase de la ganancia de lazo abierto del VFA
AV_VFA_mag = Ad0 / (np.sqrt(1 + (fg/fp1_VFA)**2) * np.sqrt(1 + (fg/fp2_VFA)**2))
AV_VFA_fase = -np.degrees(np.arctan(fg/fp1_VFA)) - np.degrees(np.arctan(fg/fp2_VFA))

# =============================================================================
# Analisis del amplificador CFA
# =============================================================================
# Condicion de diseño de Magnitud y Fase del lazo T
T_mag = 1
T_fase = -180 + M_phi

# Magnitud y fase de la ganancia de lazo cerrado
K_fase = 0
AVf_CFA_mag = T_mag / (K * AV_VFA_mag)
AVf_CFA_fase = T_fase - K_fase - AV_VFA_fase

fp_CFA = fg / np.tan(np.deg2rad(-AVf_CFA_fase))  # Polo del CFA (Hz)

AVf_CFA_DC = AVf_CFA_mag * np.sqrt(1 + (fg/fp_CFA)**2)  # Ganancia DC de lazo cerrado del CFA (veces)

# Red de realimentacion
R2 = 1/(2 * np.pi * fp_CFA * CT)
R1 = R2/(AVf_CFA_DC - 1)

fp_3dB = fg / 0.644 # Ancho de banda a -3dB (Hz)

# =============================================================================
# Resultados numericos
# =============================================================================
print("="*50)
print("Analisis VFA")
print("-"*50)
print(f"Magnitud |AV_VFA|= {AV_VFA_mag:.3f}")
print(f"Fase ∠AV_VFA= {AV_VFA_fase:.2f}°")
print("="*50)
print("Analisis CFA")
print("-"*50)
print(f"Magnitud |AVf_CFA|= {AVf_CFA_mag:.1f}")
print(f"Fase ∠AVf_CFA= {AVf_CFA_fase:.2f}°")
print(f"Polo fp_CFA= {fp_CFA/1e6:.1f} MHz")
print("-"*50)
print(f"Ganancia DC AVf_CFA= {AVf_CFA_DC:.1f}")
print(f"R2= {R2:.0f} Ω")
print(f"R1= {R1:.2f} Ω")
print("-"*50)
print(f"Ancho de banda (-3dB)= {fp_3dB/1e6:.1f} MHz")
```

```python
# =============================================================================
# Respuesta al escalon
# =============================================================================
V_pico = 122.55e-3
V_regimen = 97.96e-3

OS = (V_pico - V_regimen) / V_regimen * 100     # Sobrepaso (overshoot) porcentual

# Factor de amortiguamiento
ln_os = np.log(OS / 100)
zeta = -ln_os / np.sqrt(np.pi**2 + ln_os**2)

# Margen de fase (grados)
Mp = np.arctan((2 * zeta) / np.sqrt(np.sqrt(1 + 4 * zeta**4) - 2 * zeta**2)) * (180 / np.pi)

print("="*50)
print("Respuesta al escalon")
print("-"*50)
print(f"Sobrepaso (OS) = {OS:.2f} %")
print(f"Zeta = {zeta:.3f}")
print(f"Margen de fase (Mp) = {Mp:.2f}°")
```
